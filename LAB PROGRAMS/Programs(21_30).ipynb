Experiment 21: Sharpening using Laplacian Mask with Diagonal Neighbors

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img = cv2.imread("image1.jpg")

kernel = np.array([[1, 1, 1],
                   [1, -8, 1],
                   [1, 1, 1]])

sharpened = cv2.filter2D(img, -1, kernel)

print("Sharpened Image")
cv2_imshow(sharpened)

cv2.imwrite("Sharpened_Diagonal.jpg", sharpened)
```

Experiment 22: Sharpening using Laplacian Mask with Positive Center Coefficient

Experiment 23: Sharpening using Unsharp Masking

```python
import cv2
from google.colab.patches import cv2_imshow

img = cv2.imread("image.jpg")

blur = cv2.GaussianBlur(img, (9, 9), 10)

sharpened = cv2.addWeighted(img, 1.5, blur, -0.5, 0)

print("Unsharp Masking")
cv2_imshow(sharpened)

cv2.imwrite("Unsharp_Masking.jpg", sharpened)
```

Experiment 24: Sharpening using High-Boost Filtering

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img = cv2.imread("image1.jpg")

blur = cv2.GaussianBlur(img, (5, 5), 0)

mask = cv2.subtract(img, blur)

high_boost = cv2.addWeighted(img, 1.5, mask, 1.0, 0)

print("High Boost Image")
cv2_imshow(high_boost)

cv2.imwrite("High_Boost.jpg", high_boost)
```

Experiment 26: Insert Watermark into an Image using OpenCV

```python
import cv2
from google.colab.patches import cv2_imshow

img = cv2.imread("image.jpg")
logo = cv2.imread("image1.jpg")

logo = cv2.resize(logo, (100, 100))

h, w = logo.shape[:2]

roi = img[-h:, -w:]

result = cv2.addWeighted(roi, 0.7, logo, 0.3, 0)

img[-h:, -w:] = result

print("Watermarked Image")
cv2_imshow(img)

cv2.imwrite("watermarked.jpg", img)
```

Experiment 27: Cropping, Copying and Pasting an Image

```python
import cv2
from google.colab.patches import cv2_imshow

img = cv2.imread("image1.jpg")

h, w = img.shape[:2]

crop = img[h//4:h//2, w//4:w//2]

ch, cw = crop.shape[:2]

img[0:ch, 0:cw] = crop

print("Crop, Copy and Paste")
cv2_imshow(img)

cv2.imwrite("crop_copy_paste.jpg", img)
```

Experiment 28: Find the Boundary of an Image using Convolution Kernel

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img = cv2.imread("image1.jpg")

kernel = np.array([[-1,-1,-1],
                   [-1, 8,-1],
                   [-1,-1,-1]])

boundary = cv2.filter2D(img, -1, kernel)

print("Boundary Detection")
cv2_imshow(boundary)

cv2.imwrite("boundary.jpg", boundary)
```

Experiment 29: Morphological Operations using Erosion Technique

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img = cv2.imread("image.jpg", 0)

kernel = np.ones((5,5), np.uint8)

eroded = cv2.erode(img, kernel, iterations=1)

print("Erosion")
cv2_imshow(eroded)

cv2.imwrite("erosion.jpg", eroded)
```

Experiment 30: Morphological Operations using Dilation Technique

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img = cv2.imread("image1.jpg", 0)

kernel = np.ones((5,5), np.uint8)

dilated = cv2.dilate(img, kernel, iterations=1)

print("Dilation")
cv2_imshow(dilated)

cv2.imwrite("dilation.jpg", dilated)
```
