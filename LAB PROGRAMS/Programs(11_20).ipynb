Experiment 11: Affine Transformation on an Image

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img = cv2.imread("image1.jpg")

rows, cols = img.shape[:2]

M = np.float32([[1, 0, 100],
                [0, 1, 50]])

affine = cv2.warpAffine(img, M, (cols, rows))

print("Affine Transformation")
cv2_imshow(affine)

cv2.imwrite("Affine_Transformation.jpg", affine)
```

Experiment 12: Perspective Transformation on an Image

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img = cv2.imread("image1.jpg")

rows, cols = img.shape[:2]

src = np.float32([[0,0],[cols-1,0],[0,rows-1],[cols-1,rows-1]])
dst = np.float32([[0,0],[cols-1,0],[int(0.3*cols),rows-1],[int(0.7*cols),rows-1]])

M = cv2.getPerspectiveTransform(src, dst)

perspective = cv2.warpPerspective(img, M, (cols, rows))

print("Perspective Transformation")
cv2_imshow(perspective)

cv2.imwrite("Perspective.jpg", perspective)
```

Experiment 13: Perspective Transformation on Video

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

cap = cv2.VideoCapture("video1.mp4")

ret, frame = cap.read()

if ret:

    h, w = frame.shape[:2]

    src = np.float32([[50,50],[w-50,50],[50,h-50],[w-50,h-50]])
    dst = np.float32([[0,0],[w,0],[100,h],[w-100,h]])

    M = cv2.getPerspectiveTransform(src,dst)

    output = cv2.warpPerspective(frame,M,(w,h))

    print("Perspective Video Frame")
    cv2_imshow(output)

    cv2.imwrite("Perspective_Frame.jpg",output)

cap.release()
```

Experiment 14: Transformation using Homography Matrix

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img = cv2.imread("image.jpg")

rows, cols = img.shape[:2]

src = np.float32([[0,0],[cols-1,0],[0,rows-1],[cols-1,rows-1]])
dst = np.float32([[0,0],[cols-1,0],[0,int(rows*0.7)],[cols-1,int(rows*0.7)]])

H, status = cv2.findHomography(src,dst)

output = cv2.warpPerspective(img,H,(cols,rows))

print("Homography Transformation")
cv2_imshow(output)

cv2.imwrite("Homography.jpg",output)
```

Experiment 15: Transformation using Direct Linear Transformation (DLT)

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img = cv2.imread("image1.jpg")

rows, cols = img.shape[:2]

src = np.float32([[0,0],[cols-1,0],[0,rows-1],[cols-1,rows-1]])
dst = np.float32([[30,30],[cols-30,20],[50,rows-40],[cols-40,rows-20]])

M = cv2.getPerspectiveTransform(src,dst)

output = cv2.warpPerspective(img,M,(cols,rows))

print("DLT Transformation")
cv2_imshow(output)

cv2.imwrite("DLT.jpg",output)
```

Experiment 19: Edge Detection using Sobel Matrix along XY-axis

```python
import cv2
from google.colab.patches import cv2_imshow

img = cv2.imread("image.jpg", 0)

sobelx = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=3)
sobely = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=3)

edges = cv2.addWeighted(sobelx, 0.5, sobely, 0.5, 0)

edges = cv2.convertScaleAbs(edges)

print("Sobel XY")
cv2_imshow(edges)

cv2.imwrite("sobel_xy.jpg", edges)
```

Experiment 20: Sharpening using Laplacian Mask with Negative Center Coefficient

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img = cv2.imread("image1.jpg")

kernel = np.array([[0, 1, 0],
                   [1, -4, 1],
                   [0, 1, 0]])

sharpened = cv2.filter2D(img, -1, kernel)

print("Sharpened Image")
cv2_imshow(sharpened)

cv2.imwrite("Sharpened_Image.jpg", sharpened)
```
