EXP_1 Image to a Grayscale image

```python
import cv2
from google.colab.patches import cv2_imshow

img = cv2.imread("image.jpg")

if img is None:
    print("Error: Image not found!")
else:

    gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)


    cv2_imshow(gray_img)

    cv2.imwrite("gray_image.jpg", gray_img)

    print("Image saved successfully.")
```

EXP_2 Read an Image and Apply Gaussian Blur

```python
import cv2
from google.colab.patches import cv2_imshow

image = cv2.imread("image.jpg")

if image is None:
    print("Error: Image not found!")
else:

    blur = cv2.GaussianBlur(image, (5, 5), 10)


    cv2_imshow(blur)


    cv2.imwrite("blurred_image.jpg", blur)

    print("Blurred image saved successfully.")
```

EXP_3 Read an Image and Detect Edges using the Canny Function

```python
import cv2
from google.colab.patches import cv2_imshow

img = cv2.imread("image1.jpg")

if img is None:
    print("Error: Image not found!")
else:

    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)


    edges = cv2.Canny(gray, 100, 200)

    cv2_imshow(edges)
    cv2.imwrite("canny_edges.jpg", edges)

    print("Canny edge image saved successfully.")
```

EXP_4 Read an Image and Perform Image Dilation using the Dilate Function

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img = cv2.imread("image1.jpg", cv2.IMREAD_GRAYSCALE)

if img is None:
    print("Error: Image not found!")
else:

    kernel = np.ones((5, 5), np.uint8)

    dilated_img = cv2.dilate(img, kernel, iterations=1)

    cv2_imshow(dilated_img)
    cv2.imwrite("dilated_image.jpg", dilated_img)

    print("Dilated image saved successfully.")
```

EXP_5 Read an Image and Perform Image Erosion using the Erode Function

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img = cv2.imread("image1.jpg")

if img is None:
    print("Error: Image not found!")
else:

    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (5, 5))

    eroded_img = cv2.erode(gray, kernel, iterations=1)

    cv2_imshow(eroded_img)

    cv2.imwrite("eroded_image.jpg", eroded_img)

    print("Eroded image saved successfully.")
```

EXP_6 Read Captured Video and Display the Video in Slow Motion and Fast Motion

```python
import cv2
from google.colab.patches import cv2_imshow

cap = cv2.VideoCapture("video1.mp4")

frame_count = 0

while cap.isOpened():
    ret, frame = cap.read()

    if not ret:
        break

    frame_count += 1

    if frame_count % 30 == 0:
        slow_frame = cv2.resize(frame, None, fx=0.5, fy=0.5)
        fast_frame = cv2.resize(frame, None, fx=2.0, fy=2.0)

        print(f"Frame {frame_count} - Slow Motion")
        cv2_imshow(slow_frame)

        print(f"Frame {frame_count} - Fast Motion")
        cv2_imshow(fast_frame)

cap.release()
print("Video processing completed.")
```

EXP_7 Capture video from web Camera and Display the video

```python
import cv2
from google.colab.patches import cv2_imshow

cap = cv2.VideoCapture("video.mp4")

print("Processing Video...")

while cap.isOpened():
    ret, frame = cap.read()

    if not ret:
        break

    print("Slow Motion Frame")
    cv2_imshow(frame)
    cv2_imshow(frame)

    ret, fast_frame = cap.read()

    if not ret:
        break

    print("Fast Motion Frame")
    cv2_imshow(fast_frame)

cap.release()

print("Video Processing Completed.")
```

EXP_9 Perform Rotation of an image to clockwise and counter clockwise direction.

```python
import cv2
from google.colab.patches import cv2_imshow

img = cv2.imread('image.jpg')

cw = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)
ccw = cv2.rotate(img, cv2.ROTATE_90_COUNTERCLOCKWISE)

cv2_imshow(cw)
cv2_imshow(ccw)
```

EXP_10 Perform moving of an image from one place to another.

```python
from google.colab.patches import cv2_imshow
import cv2
import numpy as np

img = cv2.imread("image.jpg")

M = np.float32([[1, 0, 100],
                [0, 1, 50]])

translated_img = cv2.warpAffine(img, M, (img.shape[1], img.shape[0]))

print("Original Image")
cv2_imshow(img)

print("Image after Translation (Moved 100 pixels Right and 50 pixels Down)")
cv2_imshow(translated_img)
```
